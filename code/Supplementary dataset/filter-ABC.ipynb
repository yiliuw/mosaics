## Additional filter test for ABC dataset

```r
test = test %>% rename(x = x_section, y = y_section)
```

```r
section_filter <- function(data,sec,is_exc){
    li<- output_sample(test, sec)
    if (is_exc == TRUE){    
        type <- li$et
        data<-li$edata
    }
    else{
        type <- li$it
        data<-li$idata
    }
    pp<- create_pp(type,data)
    nn<-calc_R(pp, type, 0.2)
    conf<- calc_conf(pp, type, test, sec)
    res<-intense_qualify(pp,type, nn, conf)
    col.name<-paste('%',toString(sec), sep='')
    df<- data.frame(cluster = res$cluster, coln = round(res$conf,2))
    names(df)[names(df) == "coln"] <- col.name
    return(df)
}
```

### Excitatory cells

```r
for (n in 59:61){
    nam <- paste('df',toString(n), sep='')
    assign(nam, section_filter(test, n, TRUE))
}
```

### Inhibitory cells

```r
for (n in 59:61){
    nam <- paste('dfi',toString(n), sep='')
    assign(nam, section_filter(test, n, FALSE))
}
```

## Hypothesis testing on qualified types

```r
df<- read.csv("./filter-inhib.csv")
df1 <- replace(df, is.na(df), 0)
df1[df1<=0.5] <- NA
df1<-df1[rowSums(is.na(df1)) != ncol(df1)-1, ]
row.names(df1) <- NULL
types1 = df1$cluster
types1
```

```r
sections1<- apply(df1[,-1], 1, function(i) colnames(df1[,-1])[ !is.na(i) ]) ## qualified colns
sections1<- sapply(sections1, function(x){as.numeric(gsub("X.", "", x))})
sections1
```

```r
source("./scale.R")
options(warn=-1)
```

```r
## Create point patterns
pp1<-create_pp(test, NULL,sections1[[1]], types1[1])
pp2<-create_pp(test, NULL,sections1[[2]], types1[2])
pp3<-create_pp(test, NULL,sections1[[3]], types1[3])
```

```r
npp <-npoints(pp1[[1]])
Wpp <-Window(pp1[[1]])
pp.base <-rSSI(npp, r=0.01, Wpp, nsim=199)
dclf.test(pp1[[1]], Lest, simulate=pp.base, nsim=199, rinterval=c(0,0.05), alternative='less',verbose=FALSE)$p.value
```

```r
npp <-npoints(pp2[[1]])
Wpp <-Window(pp2[[1]])
pp.base <-rSSI(npp, r=0.01, Wpp, nsim=199)
dclf.test(pp2[[1]], Lest, simulate=pp.base, nsim=199, rinterval=c(0,0.05), alternative='less',verbose=FALSE)$p.value
```

```r
npp <-npoints(pp3[[1]])
Wpp <-Window(pp3[[1]])
pp.base <-rSSI(npp, r=0.01, Wpp, nsim=199)
dclf.test(pp3[[1]], Lest, simulate=pp.base, nsim=199, rinterval=c(0,0.1), alternative='less',verbose=FALSE)$p.value
```
